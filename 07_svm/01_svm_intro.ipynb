### 💡 SVM (Support Vector Machine) : 서포트 벡터 머신

`SVM`은 `KNN`, `Decision Tree`, `Radom Forest`, `Logistic Regression`와 같은 분류문제에 사용할 수 있는 알고리즘이며, **서로 다른 클래스를 구분하는 결정 결계를 찾는 알고리즘**이다.

결정 경계를 찾는데 SVM에서 나오는 중요한 점이 **마진(Margin)**이다.

마진은 결정 경계와 가장 가까운 학습 데이터 사이의 거리를 의미한다.

SVM은 기본적으로 이 마진폭을 최대한 크게 만드는 결정 경계를 찾는것이 목적이다.



#### 1️⃣ 초평면(Hyperplane)
초평면은 데이터를 분류하는 결정 경계를 의미한다.

- 입력 변수가 2개이면 직선
- 입력 변수가 3개이면 평면
- 입력 변수가 더 많으면 고차원 초평면

그냥 **단순히 클래스를 나누는 경계**라는 뜻이다.

#### 2️⃣ 마진(Margin)
마진은 결정 경계와 가장 가까운 데이터 사이의 여유 공간이다.

SVM은 마진을 크게 확보하려한다. 

#### 3️⃣ 서포트 벡터(Support Vector)
서포트 벡터는 결정 경계를 정하는데 중요한 역할을 하는 학습 데이터이다.

모든 데이터가 경계에 동일한 영향을 주는 것은 아니고, 경계 주변의 중요한 데이터들이 SVM의 결정 경계를 결정하는데 핵심적인 역할을 한다.

```python
# svm 학습해보기
import numpy as np
from sklearn.svm import SVC

# 공부 시간
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# 합격 여부
y = np.array([0, 0, 0, 1, 1, 1])

# 선형 SVM 모델
# SVC() : SVM 분류 모델 생성
# kernel="linear" : 선형 결정 경계 사용
model = SVC(kernel="linear")

# 학습
model.fit(X, y)

# 새로운 데이터 예측
X_new = np.array([[2], [4], [5]])

print(f"예측 결과 : {model.predict(X_new)}")
```

```python
# 서포트 벡터 확인해보기
# 서포트 벡터 : 결정 경계를 정하는데 중요하게 기여한 데이터

# support_vectors_ : 서포트 벡터 데이터
print("서포트 벡터 :")
print(model.support_vectors_)

# support_ : 원본 학습 데이터에서의 인덱스
print("서포트 벡터 인덱스 :")
print(model.support_)

# n_support_ : 클래스별 서포트 벡터 개수
print("클래스별 서포트 벡터 개수 :")
print(model.n_support_)
```
