### 💡 Magin & Support Vector

#### 1️⃣ Margin이란
마진은 결정 경계 양쪽에 확보하려는 공간이다. SVM은 두 클래스를 구분하는 것 뿐만 아니라, 가장 가까운 학습 데이터들과 결정 경계 사이에 가능한 한 넓은 공간을 확보하려고 한다.

결정 경계는 두 클래스 사이에 위치하며, 결정 경계를 기준으로 양쪽으로 여유 공간이 있다.

이 여유 공간을 크게 확보하면 새로운 데이터가 경계 근처에 들어와도 작은 변화에 덜 민감하게 분류할 가능성이 높아진다.

하지만, 마진이 크다고 항상 예측이 좋은건 아니다. 실제 데이터에서는 분류 오류도 함께 고려해야 한다.

#### 2️⃣ Support Vector란
서포트 벡터는 SVM의 결정 경계와 마진을 정하는데 직접적인 영향을 주는 학습 데이터이다.

모든 데이터가 결정 경계에 동일하게 영향을 주지는 않을 것이다.

- 결정 경계에서 멀리 떨어진 데이터일수록 결정 경계에 미치는 영향이 작다.
- 결정 경계 근처의 서포트 벡터는 경계와 마진을 결정하는데 중요한 역활을 한다.

한줄 요약:

> 서포트 벡터는 결정 경계를 정하는 중요한 데이터이고, 마진은 그 경계 양쪽에 확보하려는 여유 공간

```python
# SVM 모델 학습 및 서포트 벡터 확인
import numpy as np
from sklearn.svm import SVC

X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

model = SVC(kernel="linear")
model.fit(X, y)

print("서포트 벡터:")
print(model.support_vectors_)

print("서포트 벡터 인덱스:")
print(model.support_)

print("서포트 벡터 갯수:")
print(model.n_support_)
```

#### 3️⃣ decision_function()
svm에서는 `decision_function()`을 통해 새로운 데이터가 결정 경계의 어느쪽에 위치하는지 확인할 수 있다.

svm의 `decision_function()`점수는 확률도 아니고, 일반적인 유클리드 거리도 아니다. 그냥 절대값이 클수록 경계에서 더 멀리 떨어져 있다는 뜻으로 해석이 가능하다.

```python
X_new = np.array([[3], [4], [4.5], [5], [6]])

print("결정 함수 점수:")
print(model.decision_function(X_new))

print("예측 클래스:")
print(model.predict(X_new))
```

#### 실습문제

```python
import numpy as np
from sklearn.svm import SVC

X = np.array([2, 3, 4, 5, 6, 7, 8, 9]).reshape(-1, 1)
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# 1. 선형 SVM 모델 생성 및 학습
model = SVC(kernel="linear")
model.fit(X, y)

# 2. 서포트 벡터 출력
print(model.support_vectors_)

# 3. 서포트 벡터의 원본 데이터 인덱스 출력
print(model.support_)

# 4. 클래스별 서포트 벡터 개수 출력
print(model.n_support_)

# 5. 서포트 벡터가 왜 해당 데이터인지 주석으로 설명
# -> 서포트 벡터는 결정 경계를 정하는데 주요한 역할을 한 데이터들이다.

X_new = np.array([[4], [5], [6], [7]])

# 1. 새로운 데이터의 예측 클래스 출력
print(model.predict(X_new))

# 2. decision_function() 점수 출력
print(model.decision_function(X_new))

# 3. 어떤 데이터가 결정 경계에 가장 가까운지 확인
# -> 5, 6 두개의 데이터가 결정경계에 가장 가깝다.

# 4. 점수가 음수와 양수인 경우의 의미 설명
# -> 음수일수록 결정경계의 왼쪽, 양수일수록 오른쪽에 위치하며 절대값이 클수록 결정경계로부터 멀리 떨어진 값이다.

# 1. SVM에서 마진(Margin)이란 무엇일까?
# -> 마진이란 결정경계와 서포트 벡터 사이의 공간을 의미한다.

# 2. 서포트 벡터가 결정 경계를 정하는 데 중요한 이유는 무엇일까?
# -> 결정경계는 서포트 벡터로 인해 정해지기 때문

# 3. 결정 경계에서 멀리 떨어진 데이터와 서포트 벡터는 어떤 차이가 있을까?
# -> 서포트벡터는 결정경계를 정하는데 주요한 역할을 하며, 멀리 떨어진 데이터일수록 결정경계를 정하는데 기여하는 바가 적다.

# 4. support_vectors_와 support_는 어떤 차이가 있을까?
# -> support_vecotrs : 서포트 벡터 출력
# -> support_ : 서포트 벡터의 인덱스값 출력

# 5. SVM에서 decision_function()이 반환하는 값은 확률일까?
# -> 결정 경계로부터 얼마나 멀리 떨어졌는지를 측정하는 척도
```
