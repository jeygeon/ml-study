### 💡 Distance Metric (거리 측정 척도)
거리 측정 척도란 두 데이터가 얼마나 떨어져 있는지를 수치로 나타내는 방법이다.

KNN에서는 일반적으로 다음과 같이 판단한다.
- 거리가 작다 &rightarrow; 두 데이터의 특성이 비슷하다.
- 거리가 크다 &rightarrow; 두 데이터의 특성이 다르다.
- 거리가 0이다 &rightarrow; 측정에 사용한 특성값이 동일하다.

#### 1️⃣ 유클리드 거리 (Euclidean Distance)

유클리드 거리는 두 점 사이를 직선으로 연결했을 때의 거리이다.

유클리드 거리의 공식은 다음과 같다.

$d(A, B) = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$

우리가 알고있는 피타고라스 정리와 동일하다.

````python
# 파이썬으로 구현해보기
import numpy as np

a = np.array([1, 0])
b = np.array([4, 4])

# 각 원소의 차이
diff = a - b

# 차이의 제곱
squared = diff ** 2

# 제곱의 합
sum = squared.sum()

# 합의 루트
distance = np.sqrt(sum)
print(distance)
````

````python
# numpy 에서 제공하는 np.linalg.norm()을 이용하면 간편하게 구할 수 있다.
print(np.linalg.norm(a - b))
````

#### 2️⃣ 맨해튼 거리 (Manhattan Distance)

맨해튼 거리는 각 특성의 차이에 절대값을 취한 뒤 모두 더한 거리이다.

맨해튼 거리는 제곱이나 제곱근을 사용하지 않는다.

$d(A, B) = |x_2 - x_1| + |y_2 - y_1|$

````python
# python으로 구현해보기
a = np.array([1, 0])
b = np.array([4, 4])

# 각 원소의 차이
diff = b - a

# 차이의 절대값
absoulate = np.abs(diff)

# 합계
distance = absoulate.sum()
print(distance)
````

#### 3️⃣ KNN모델에서 거리 계산 방식 설정하기

```python
metric = "manhattan"
```
으로 변경하면 된다.
