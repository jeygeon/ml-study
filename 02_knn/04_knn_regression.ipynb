### 💡 KNN Regression

KNN은 분류와 회귀 모두 가까운 K개의 이웃을 찾는 과정까지는 동일하다.

분류는 이웃을 찾은 이후에 가장 가까운 이웃들을 참고해서 범주를 예측한다면,

회귀는 이웃들의 평균을 가지고 수치를 예측한다는 차이가 있다.

```python
# KNeighborsRegressor 사용해보기
# 아파트 면적을 가지고 얼마일지 예측해보기
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.neighbors import KNeighborsRegressor

# 아파트 면적
X = np.array([
    [10], [15], [20], [24], [27], [30], [35]
])

# 아파트 거래 가격
y = np.array([
    2.0, 3.0, 4.0, 4.8, 5.4, 6.0, 8.0
])

# 모델 생성
model = KNeighborsRegressor(n_neighbors=3)

# 모델 학습
model.fit(X, y)

# 새로운 데이터 생성
# 면적이 26인 아파트
new_X = np.array([[26]])

# 가격 예측
new_y = model.predict(new_X)
print(new_y)

# 예측에 사용한 이웃 확인
distances, indices = model.kneighbors(new_X)

# 선택된 이웃들의 아파트 면적
print(X[indices])

# 선택된 이웃들의 아파트 가격
print(y[indices])

# 선택된 이웃들의 가격 평균
print(np.mean(y[indices]))

# 모델의 예측 결과
print(new_y)
```

‼️ `KNN` 회귀의 특징은 다음과 같이 설명할 수 있다.
- 가까운 이웃들의 목표값 평균을 이용하여 수치를 예측한다.
- 별도의 회귀식을 생성하지 않고 수치를 예측할 수 있다.
- 선형 관계뿐만 아니라 비선형 관계도 표현할 수 있다.
- 기본 KNN 회귀의 예측값은 선택된 이웃들의 목표값이 동일하면 변화하지 않으므로 계단 형태의 예측 그래프가 나타난다. (동일 가중치 기준)
- 훈련 데이터의 목표값 최솟값보다 작거나 최댓값보다 큰 값을 예측할 수 없다. (기본 평균 방식 기준)
