### 💡 KNN이란?
`KNN`은 **K-Nearest Neighbors**의 약자로, 새로운 데이터가 들어왔을 때 기존 데이터 중 가장 비슷한 K개의 데이터를 참고해서 결과를 예측하는 머신러닝 알고리즘이다.
- `K` : 참고할 데이터의 개수
- `Nearest` : 가장 가까운
- `Neighbors` : 이웃 데이터

쉽게 말하면 **나와 비슷한 사람들은 어떠한 결과를 가지고 있는지 확인하고, 그것을 바탕으로 나의 결과를 예측**하는 방법이다.

아래와 같은 운동 데이터가 있다고 가정해보자

| 회원 | 운동 경력(년) | 주간 운동 횟수 | 숙련도 |
| --- | --- | --- | --- |
| A | 1 | 1 | 초급자 |
| B | 1 | 1 | 초급자 |
| C | 2 | 2 | 초급자 |
| D | 3 | 3 | 숙련자 |
| E | 3 | 4 | 숙련자 |

머신러닝 모델은 다음과 같은 정보를 이용한다.
- **Featrue**($X$) : 운동 경력, 주간 운동 횟수
- **Target**($y$) : 숙련도

#### 1️⃣ KNN의 동작원리
`KNN`은 **거리 계산**, **이웃 선택**, **결과 예측** 이런 추론 과정을 거친다.

1. 신규 회원과 기존 회원 사이의 거리를 계산한다. (이 때 거리 계산에는 유클리드 거리를 사용한다.)
2. K가 3이라면 가장 가까운 회원 3명을 선택한다.
3. 선택된 이웃 중 가장 많이 등장하는 범주를 선택한다. (이러한 방식을 다수결 투표라고한다.)

#### 2️⃣ KNN의 분류와 회기
`KNN`은 **분류**와 **회귀** 모두 사용할 수 있는 **지도학습**이다.

두 방식 모두 가까운 이웃을 찾는 과정은 동일하지만, 최종 결과를 결정하는 방법이 다르다.

분류는 이웃 데이터의 범주 중 가장 많이 등장한 범주를 선택하고, 회귀는 기본적으로 이웃 데이터의 수치 평균을 계산한다.

| 구분 | KNN 분류 | KNN 회귀 |
| --- | --- | --- |
| 모델 | KNeighborsClassifier | KNeighborsRegressor |
| 에측 대상 | 범주형 데이터 | 연속형 수치 데이터 |
| 예측 방법 | 다수결 | 평균 |

#### 3️⃣ Lazy Learning
`KNN`은 `Lazy Learning`이다. `Eager Learning`과 `Lazy Learning`의 차이는 아래와 같은 예를 들며 설명할 수 있다.

- 학생 A (Eager Learning) : 시험 전에 교재를 읽고 문제를 푸는 방법과 규칙을 학습한다. 시험을 볼 때는 학습한 규칙을 이용하여 문제를 해결한다.
- 학생 B (Lazy Learning) : 시험 전에 교재를 준비해둔다. 시험 문제가 나오면 그 때 교재를 확인해서 비슷한 문제를 찾아서 문제를 해결한다.

`KNN`은 학생 B와 비슷하다. 일반적인 머신러닝 모델은 학습 과정에서 미리 데이터의 패턴이나 규칙을 학습하지만, `KNN`은 별도의 예측 공식을 학습하지않고 기존의 데이터를 계속 보관한 뒤,

새로운 데이터가 들어오면 주변 이웃을 찾아 결과를 예측한다.

`KNN`은 일반적으로 학습 속도가 빠른편이지만, 데이터가 많아질수록 기존 데이터를 저장하기 때문에 점점 속도가 느려질 수 있다.

```python
# KNN 구현 맛보기
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

# Feature: 운동 경력, 주간 운동 횟수
X = np.array([
    [1, 1],
    [1, 2],
    [2, 2],
    [3, 3],
    [4, 4],
    [5, 5]
])

# Target: 0 = 초급자, 1 = 숙련자
y = np.array([0, 0, 0, 1, 1, 1])

# knn 모델 생성
# n_neighbors=3 : 새로운 데이터 예측할 때 가장 가까운 데이터 3개를 참고
model = KNeighborsClassifier(n_neighbors=3)

model.fit(X, y)

new_data = np.array([[3, 6]])
pred = model.predict(new_data)
print(pred)

# 어떤 이웃을 참고했는지도 확인 가능하다.
distance, indices = model.kneighbors(new_data)
print("거리 :", distance)
print("인덱스 :", indices)
print("이웃의 정답 :", y[indices])
```

### 실습

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

X = np.array([
    [1, 5],
    [2, 10],
    [3, 15],
    [5, 25],
    [6, 30],
    [7, 35],
    [8, 40]
])

y = np.array([0, 0, 0, 1, 1, 1, 1])

new_student = np.array([[4, 20]])

# 1. 데이터 구조 확인
print(X.shape)
print(y.shape)

# 2. K=3 모델 생성
knn_k3 = KNeighborsClassifier(n_neighbors=3)

# 3. 모델 학습
knn_k3.fit(X, y)

# 4. 예측
k3_result = knn_k3.predict(new_student)

# 5. 이웃의 거리와 인덱스 확인
distances, indices = knn_k3.kneighbors(new_student)

print("거리:", distances)
print("인덱스:", indices)

# 6. 이웃의 실제 합격 여부 확인
print("이웃의 정답:", y[indices])

# 7. K=5 모델 생성 및 학습
knn_k5 = KNeighborsClassifier(n_neighbors=5)
knn_k5.fit(X, y)

k5_result = knn_k5.predict(new_student)

# 8. 예측 결과 비교
print("K3 결과:", k3_result)
print("K5 결과:", k5_result)
```
