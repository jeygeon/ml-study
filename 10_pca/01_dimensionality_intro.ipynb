### 💡 Dimensionality : 차원

머신러닝에서 **차원이란 하나의 데이터를 표현하기 위해 사용하는 입력 특성(Feature)의 개수를 의미**한다.

예를들어 집값을 예측한다고 가정했을 때
- 1차원 &rightarrow; 면적
- 2차원 &rightarrow; 면적, 방 개수
- 3차원 &rightarrow; 면적, 방 개수, 건축 연도
- 4차원 &rightarrow; 면적, 방 개수, 건축 연도, 역까지 거리

이렇듯 입력 특성이 하나씩 추가될 때마다 차원도 하나씩 증가한다.

> ❗️ 주의할점은 2차원의 배열로 저장되어있다고, 머신러닝에서 2차원의 데이터라는 의미는 아니다. **PCA에서 말하는 차원은 입력 특성의 개수를 의미한다.**

#### 1️⃣ 차원이 증가하는 이유

머신러닝에서 차원이 증가하는 이유는 대표적으로 세 가지가 있다.

1. 새로운 정보 추가

    - 말 그대로 새로운 Feature가 추가는 경우이다.

2. 파생 변수 생성

    - 기존 특성을 활용하여 새로운 특성을 생성하는 경우이다. ex) 사이즈를 가지고 면적을 계산

3. 범주형 변수 변환

    - 원핫 인코딩을 사용하면 하나의 범주형 변수가 여러개의 특성으로 변환될 수 있다.

#### 2️⃣ Curse of Dimensionality : 차원의 저주

차원의 저주는 데이터의 차원이 증가하면서 데이터 분석과 머신러닝에서 발생하는 어려움을 의미한다.

일닫 거리 기반 알고리즘의 성능 문제가 발생할 수 있다.

KNN이나 K-Means같은 경우 데이터 사이의 거리를 중요하게 판단한다.

하지만 관련없는 특성이 많이 추가되면 의미있는 특성보다 불필요한 특성이 거리 계산에 영향을 줄 수 있다.

차원 증가의 주요 문제를 정리하자면 아래와 같다.

1. **데이터 희소성**
    
    - 넓어진 공간에 데이터가 상대적으로 부족해짐

2. **계산량 증가**

    - 학습 및 예측 비용 증가

3. **메모리 사용량 증가**

    - 저장해야 하는 특성이 많아짐

4. **과적합 가능성**

    - 불필요한 패턴까지 학습할 수 있음

5. **거리 계산 문제**

    - 거리 기반 알고리즘의 구분 능력이 약해질 수 있음

따라서 특성을 추가할 때는 검증 데이터의 예측 성능이 개선되는 효과, 특성 데이터에서만 성능이 좋아진것은 아닌지 확인하는 안정성 등 고려해야할 사항이 있다.

#### 3️⃣ Dimensionality Reduction : 차원 축소

**차원 축소는 데이터의 중요한 정보를 가능한 한 유지하면서 입력 특성의 개수를 줄이는 과정**이다.

차원 축소는 크게 두 가지 방법으로 나눌 수 있다.
- **Feature Selection** : 기존 특성 중 중요한 특성만 선택
- **Feature Extraction** : 기존 특성을 변환하여 새로운 특성 생성

`PCA`는 Feature Extraction에 해당하는 대표적인 차원 축소 알고리즘이다.
