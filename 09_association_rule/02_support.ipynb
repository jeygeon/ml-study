### 💡 Support

support란 전체 거래 중 특정 항목 또는 항목 집합이 등장하는 비율을 의미한다.

$Support(A \rightarrow B)=\frac{A와\ B가\ 함께\ 등장한\ 거래\ 수}{전체\ 거래\ 수}$

예를 들어
- 전체거래 : 100$
- 빵 구매 거래 : 40
- 버터 구매 거래 : 30
- 빵과 버터 동시 구매 거래 : 20

라고 한다면

$Support(빵 \rightarrow 버터)=\frac{20}{100}=0.2$

지지도는 0.2, 즉 20%이다.

`support`는 0과 1사이의 값을 가지며, 0은 한번도 등장하지 않고 1로 갈수록 모든 거래에서 등장했다고 볼 수 있다.

`support`가 높다는 것은 해당 `itemset`이 전체 거래에서 자주 등장한다는 의미이다.

#### 1️⃣ 단일 항목과 여러 항목의 Support

support는 반드시 두 개의 상품을 대상으로 계산되는 것은 아니다. 단일 상품이 될 수도 있고 두개, 세개 이상의 상품이 될 수도 있다.

```python
transactions = [
    ["우유", "빵", "계란"],
    ["우유", "빵"],
    ["빵", "계란"],
    ["우유", "계란"],
    ["우유", "빵", "계란"]
]
```

위의 Itemset으로 support를 확인해보자면 아래와 같다.
| Itemset | 등장 횟수 | Support |
| --- | --- | --- |
| 우유 | 4 | 0.8 |
| 빵 | 4 | 0.8 |
| 계란 | 4 | 0.8 |
| 우유, 빵 | 3 | 0.6 |
| 우유, 계란 | 3 | 0.6 |
| 빵, 계란 | 3 | 0.6 |
| 우유, 빵, 계란 | 2 | 0.4 |

여기서 중요한 특징은 항목 집합에 새로운 항목이 추가될수록 반드시 support 수치는 기존 값이랑 같거나 작아진다는 점이다.

$Support(A,B,C)\leq Support(A,B)\leq Support(A)$

조건이 추가될수록 모든 조건을 만족하는 거래의 수는 작아지기 떄문이다.

#### 2️⃣ Support의 방향성

support는 규칙의 방향이 바뀌어도 값은 동일하다. 즉 방향성이 없다는 의미이다.

$Support(A \rightarrow B)=Support(B \rightarrow A)$

우유와 빵의 동시 구매 Itemset이나, 빵과 우유의 동시 구매 Itemset의 갯수는 같기 때문이다.

````python
import numpy as np
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder

transactions = [
    ["우유", "빵", "계란"],
    ["우유", "빵"],
    ["빵", "계란"],
    ["우유", "계란"],
    ["우유", "빵", "계란"]
]

encoder = TransactionEncoder()

encoded_item = encoder.fit_transform(transactions)

df = pd.DataFrame(encoded_item, columns=encoder.columns_)

print(df)

# 단일 상품의 support 계산
print(f"계란의 support : {df["계란"].mean()}")

# 전체 상품의 support 계산
print(f"전체 상품의 support : {df.mean()}")

# 두 상품의 support 계산
print(f"계란과 빵의 support : {(df["계란"] & df["빵"]).mean()}")
````

#### 3️⃣ Minimum Support : 최소 지지도

예를 들어 쇼핑몰에 100개의 상품이 존재한다고 했을 때 나올 수 있는 support의 가짓수는 매우 많다.

이 때 거의 등장하지 않는 상품 조합까지 모두 분석할 필요는 없을 것이다.

이럴 때 **최소 지지도**를 사용한다.
```python
min_support = 0.5
```

전체 거래에서 50%이상 등장한 Itemset만 선택하겠다는 의미이다.

#### 실습문제

````python
import numpy as np
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder

transactions = [
    ["노트북", "마우스", "키보드"],
    ["노트북", "마우스"],
    ["모니터", "키보드"],
    ["노트북", "모니터", "마우스"],
    ["마우스", "키보드"],
    ["노트북", "마우스", "키보드"],
    ["모니터", "마우스"],
    ["노트북", "키보드"],
    ["노트북", "모니터"],
    ["노트북", "마우스", "모니터", "키보드"]
]

encoder = TransactionEncoder()

encoded_item = encoder.fit_transform(transactions)

df = pd.DataFrame(encoded_item, columns=encoder.columns_)

# 1. 노트북의 Support 출력
print(df["노트북"].mean())

# 2. 마우스의 Support 출력
print(df["마우스"].mean())

# 3. 키보드의 Support 출력
print(df["키보드"].mean())

# 4. 모니터의 Support 출력
print(df["모니터"].mean())

# 5. 전체 상품별 Support를 한 번에 계산
# 결과를 item_support 변수에 저장하고 출력
item_support = df.mean()
print(item_support)

# 6. Support가 가장 높은 상품 이름과 지지도 출력
# 공동 1위가 존재하면 모두 출력
max_support = item_support.max()
print(item_support[item_support == max_support])

# 7. Support가 가장 낮은 상품 이름과 지지도 출력
# 공동 최하위가 존재하면 모두 출력
min_support = item_support.min()
print(item_support[item_support == min_support])

# 8. 노트북과 마우스의 Support 출력
print((df["노트북"] & df["마우스"]).mean())

# 9. 마우스와 키보드의 Support 출력
print((df["키보드"] & df["마우스"]).mean())

# 10. 노트북과 모니터의 Support 출력
print((df["노트북"] & df["모니터"]).mean())

# 11. 노트북, 마우스, 키보드의 Support 출력
print((df["노트북"] & df["마우스"] & df["키보드"]).mean())

# 12. 노트북, 마우스, 모니터, 키보드의 Support 출력
print((df["노트북"] & df["마우스"] & df["키보드"] & df["모니터"]).mean())

itemsets = [
    ["노트북"],
    ["마우스"],
    ["키보드"],
    ["모니터"],
    ["노트북", "마우스"],
    ["노트북", "키보드"],
    ["마우스", "키보드"],
    ["노트북", "마우스", "키보드"]
]

# 15. 결과를 저장할 빈 리스트 results 생성


# 16. 반복문을 이용하여 itemsets의
# 8개 항목 집합에 대한 Support 계산
# df[items].all(axis=1).mean() 활용


# 17. 반복문 내부에서 각 항목 집합과 Support를
# 딕셔너리로 구성하여 results에 추가
# 딕셔너리 키: Itemset, Support


# 18. results를 DataFrame으로 변환
# 변수명: support_df
# 예상 크기: (8, 2)


# 19. support_df 전체 출력


# 20. Support를 기준으로 내림차순 정렬하여 출력
````
