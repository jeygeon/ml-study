### 💡 Lift : 향상도

Lift는 A가 등장했을 때 B가 등장할 확률이 평소 B가 등장할 확률보다 몇 배 높은지 나타내는 지표이다.

$Lift(A\rightarrow B)=\frac{Confidence(A\rightarrow B)}{Support(B)}$

예제를 살펴보면
- 전체 거래 : 100
- 노트북 구매 거래 : 20
- 생수 구매 거래 : 90
- 노트북과 생수 동시 거래 : 18

1. **Confidence 계산**

    $Confidence(노트북\rightarrow 생수)=\frac{18}{20}=0.9$

    노트북을 구매한 거래 중 90%에서 생수를 함께 구매했다. Confidence만 보면 두 상품의 연관성이 매우 강해보인다.

2. **Lift 계산**

    일단 support를 먼저 계산한 뒤

    $Support(생수)=\frac{90}{100}=0.9$

    Lift를 계산하면 아래와 같이 나온다.

    $Lift(노트북\rightarrow 생수)=\frac{0.9}{0.9}=1.0$

위 두개의 결과로 알 수 있는 사실은 전체 거래에서 생수 구매 비율은 90%이다. 그리고 노트북을 구매했을 때 생수 구매 비율도 90%이다.

즉 노트북과 생수는 특별한 연관성이 나타나지 않는다.

#### 1️⃣ Lift 값의 해석
Lift값은 아래처럼 해석이 가능하다.

| Lift | 해석 |
| --- | --- |
| 1 초과 | A가 등장할 때 B의 등장 확률이 높아짐 |
| 1 | A와 B가 독립적 |
| 1 미만 | A가 등장할 때 B의 등장 확률이 낮아짐 |

1. Lift > 1

    ```python
    Confidence(A → B) = 0.8
    Support(B)        = 0.4
    ```

    $Lift(A\rightarrow B)=\frac{0.8}{0.4}=2.0$

    A가 등장한 거래에서는 B의 등장 확률이 평소보다 2배 높다는 의미이다.

2. Lift = 1

    ```python
    Confidence(A → B) = 0.4
    Support(B)        = 0.4
    ```

    $Lift(A\rightarrow B)=\frac{0.4}{0.4}=1.0$

    A와 B와 등장 여부가 서로 독립적이라는 의미이다.

3. Lift < 1

    ```python
    Confidence(A → B) = 0.2
    Support(B)        = 0.4
    ```

    $Lift(A\rightarrow B)=\frac{0.2}{0.4}=0.5$

    A가 등장한 거래에서는 B의 등장 확률이 평소의 절반이라는 의미이다.

#### 2️⃣ Lift의 한계
Lift가 높다고 무조건 좋은 연관 규칙이란 것은 아니다.

만약 전체 거래가 10,000건인데 A와 B 각각의 거래수가 단 2건씩만 있고 A와 B 동시구매 거래수가 1건인 상황을 가정해보자면

Lift를 계산해봤을때 2,500이 나오게 된다.

Lift는 매우 큰 수치로 나타났지만 실제로 두 상품을 함께 구매한 거래는 단 1건이다.

따라서 아래와 같은 3가지 지표를 함께 고려해야 한다.

| 지표 | 확인하는 내용 |
| --- | --- |
| Support | 충분히 자주 발생하는 규칙인가? |
| Confidence | 조건이 충족되었을 때 결과가 얼마나 자주 발생하는가? |
| Lift | 평소보다 결과가 더 자주 발생하는가? |

그리고 Lift가 높다고 해서 인과관계가 존재한다는 의미도 아니니까 주의해야 한다.
