### 💡 Association Rule : 연관 규칙

`Association Rule`(연관 규칙)은 대량의 데이터에서 항목간에 숨겨진 연관성을 찾아내는 비지도학습 기법이다.

특정 항목 A가 발생했을 때 B도 함께 발생하는 경향을 찾아내는 것이 목적이다.

대표적으로 MBA(Market Basket Analysis)에 사용된다.

ex)
- 우유를 구매한 고객은 4명이다.
- 우유와 빵을 함께 구매한 고객은 3명이다.
- 우유를 구매한 고객 중 75%가 빵도 구매했다.

#### ✅ 연관 규칙의 기본 용어
1. `Transcation` : 하나의 거래
2. `Item` : 개별 항목
3. `Itemset` : 항목들의 집합
4. `Antecedent` : 선행 항목
5. `Consequent` : 후행 항목
6. `Frequent Itemset` : 빈번하게 등장하는 항목 집합

1️⃣ **Transaction**
```python
# 아래는 총 5개의 Transaction이 있다.
transactions = [
    ["우유", "빵", "계란"],
    ["우유", "빵"],
    ["빵", "계란"],
    ["우유", "계란"],
    ["우유", "빵", "계란"]
]
```

2️⃣ **Item과 Itemset**
```python
# Item
"우유"

# Itemset
{"우유", "빵"}

# Itemset
{"우유", "빵", "계란"}
```

3️⃣ **Antecedent와 Consequent**
연관 규칙은 기본적으로 아래와 같은 규칙이 있다.

$A \rightarrow B$

- Antecedent &rightarrow; 조건 항목
- Consequent &rightarrow; 결과 항목

#### ✅ 연관 규칙의 방향성
연관 규칙의 중요한 특징 중 하나는 방향성이 존재한다는 것이다.

예를 들어 전체 구매가 100건이라고 가정해보지
- 우유 구매 : 20
- 빵 구매 : 80
- 우유와 빵 동시 구매 : 18

그렇다면 아래와 같은 결과가 나올 수 있다.
- 우유를 구매한 사람들 중 빵을 함께 구매한 비율 : 18/20(90%)
- 빵을 구매한 사람들 중 우유를 함께 구매한 비율 : 18/80(22.5%)

함께 구매한 횟수는 18회로 동일하지만 기준이 되는 상품이 다르기 때문에 결과가 달라진다.

#### ✅ 연관 규칙의 핵심 평가 지표
연관 규칙에서는 아래 세 가지 핵심 지표를 사용한다.
- Support (지지도) : 전체 거래에서 해당 항목들이 함께 등장하는 비율
- Confidence (신뢰도) : A가 발생했을 때 B도 발생하는 비율
- Lift (향상도) : A와 B가 독립적으로 발생할 때와 비교한 연관성

#### ✅ 기존 머신러닝 모델과의 차이
| 비교 항목 | `Classification` | `Cluster` | `Association Rule` |
| --- | --- | --- | --- |
| 학습 유형 | 지도학습 | 비지도학습 | 비지도학습 |
| 목적 | 클래스 예측 | 유사 데이터 그룹화 | 항목 간 연관성 확인 |
| 정답 데이터 | 필요 | 불필요 | 불필요 |
| 결과 | 예측 클래스 | 군집 ID | 연관 규칙 |

- Classification : 고객이 이탈할 것인지 예측
- Clustering : 구매 성향이 비슷한 고객끼리 그룹화
- Association Rule : 함께 구매하는 상품의 관계 발견

````python
import numpy as np
import pandas as pd

transactions = [
    ["우유", "빵", "계란"],
    ["우유", "빵"],
    ["빵", "계란"],
    ["우유", "계란"],
    ["우유", "빵", "계란"]
]

# TransactionEncoder는 데이터를 Boolean 배열로 변환하는 기능이다.
from mlxtend.preprocessing import TransactionEncoder

encoder = TransactionEncoder()

# fit() : 전체 거래에서 고유 상품 종류 파악
# transform() : 각 거래를 상품 포함 여부로 변환
# fit_transform() : 두 과정을 한번에 수행
encoded_data = encoder.fit_transform(transactions)

print(encoded_data)

# 거래에서 발견한 상품 고유 목록을 확인할 수 있다.
print(encoder.columns_)

# 데이터 프레임으로 변환
df = pd.DataFrame(encoded_data, columns=encoder.columns_)
print(df)

# 상품별 구매 횟수
print(df.sum())

# 특정 상품을 함께 구매한 거래 확인
result = df[df["우유"] & df["빵"]]
print(result)
````
