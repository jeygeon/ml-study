### 💡 Confidence : 신뢰도

`Confidence`는 **A가 발생한 거래 중 B도 함께 발생한 거래의 비율**을 의미한다.

$Confidence(A \rightarrow B)=\frac{A와\ B가\ 함께\ 등장한\ 거래\ 수}{A가\ 등장한\ 거래\ 수}$

예를 들어 아래와 같은 구매 데이터가 있다고 가정해보면,
- 전체거래 : 100
- 빵 구매 : 20
- 버터 구매 : 50
- 빵과 버터 동시 구매 : 15

빵을 구매했을 때 버터도 구매하는 신뢰도를 계산하면 다음과 같다.

$Confidence(빵 \rightarrow 버터)=\frac{15}{20}=0.75$

빵을 구매한 거래 중 75%에서 버터도 함께 구매했다는 의미이다.

#### 1️⃣ Support와 Confidence의 차이

두 지표의 가장 큰 차이는 분모이다.

| 구분 | `Support` | `Confidence` |
| --- | --- | --- |
| 분자 | A와 B의 동시 등장 횟수 | A와 B의 동시 등장 횟수 |
| 분모 | 전체 거래 수 | A가 등장한 거래 수 |
| 계산 결과 | 15/100 = 0.15 | 15/20 = 0.75 |
| 방향성 | 없음 | 있음 |
| 값의 범위 | 0~1 | 0~1 |

같은 데이터를 분석하더라도 기준이 다르기 때문에 서로 다른 결과가 나온다.
- **Support**: 전체 거래의 15%에서 빵과 버터를 함께 구매했다.
- **Confidence**: 빵을 구매한 거래의 75%에서 버터도 함께 구매했다.

#### 2️⃣ Confidence의 방향성

`Confidence`는 방향에 따라 값이 달라질 수 있다.

- 빵 &rightarrow; 버터

    $Confidence(빵 \rightarrow 버터)=\frac{15}{20}=0.75$

- 버터 &rightarrow; 빵

    $Confidence(버터 \rightarrow 빵)=\frac{15}{50}=0.3$

두 상품의 동시 구매 횟수는 동일하지만 기준이 되는 상품의 구매 횟수가 다르다.

$Confidence(A \rightarrow B)\neq Confidence(B \rightarrow A)$

#### 3️⃣ Confidence와 조건부 확률

Confidence는 통계에서 조건부확률와 동일한 개념이다.

$Confidence(A \rightarrow B)=P(B|A)$

$P(B|A)$는 A가 발생했을 때 B가 발생할 확률을 의미한다.
