### 💡 KMeans Initialization

KMeans의 초기 중심점 설정 방법은 `Random Select`와 `K-Means++` 이렇게 두가지로 구분된다.

K-Means는 처음 설정한 중심점을 기준으로 데이터를 할당하면서 학습을 시작한다.

그런데 초기 중심점들이 특정 영역에 몰려있다면 학습이 비효율적으로 진행되거나 최종 군집화 결과가 좋지 않을 수 있다.

#### 1️⃣ Random Initialization

`Random Initialization`은 전체 데이터중 K개의 데이터를 무작위로 선택하여 초기 중심점으로 사용하는 방식이다.

```python
# init="random"으로 설정 가능하다.
model = KMeans(
    n_clusters=4,
    init="random",
    n_init=10,
    random_state=42
)
```

동작원리가 단순하고 구현이 간단하다는 장점이 있지만, 초기 중심점이 한곳에 몰릴 수 있고 학습 결과가 불안정할수 있다는 단점이 있다.

이러한 단점을 개선한 것이 `K-Means++`이다.

#### 2️⃣ K-Means++

`K-Means++`는 **초기 중심점들이 서로 멀리 떨어지도록 유도하는 초기화 방법**이다.

만약 군집이 4개라서 초기 중심점을 4개를 설정해야 된다면 아래와 같은 방법으로 설정된다.

1. 무작위 한개의 중심점을 선택한다.
2. 처음 고른 중심점에서 가장 먼 곳에 떨어진 데이터를 두번째 중심점으로 고른다.
3. 1번과 2번에서 고른 두 중심점에서 모두 멀리 떨어진 데이터를 중심점으로 고른다.
4. 같은 방법으로 4번째 중심점을 고른다.

> 결론적으로 한곳에 몰리지 않도록 중심점을 유도하는 것이 포인트다.

```python
# k-means++는 디폴트 값이다.
# init="k-means++"로 설정 가능하다.
model = KMeans(
    n_clusters=4,
    init="k-means++",
    n_init=10,
    random_state=42
)
```

#### 3️⃣ KMeans 하이퍼파라미터 정리
```python
model = KMeans(
    n_clusters=4,
    init="k-means++",
    n_init=10,
    max_iter=300,
    random_state=42
)
```
- `n_cluster=4` : 군집 개수 설정
- `init="k-means++"` : 초기 중심점 선택 방식
- `n_init=10` : 서로 다른 초기화로 전체 학습을 실행하는 횟수
- `max_iter=300` : 한번의 학습에서 허용되는 최대 반복 횟수

````python
# 초기화 방식 비교
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, y = make_blobs(
    n_samples=1000,
    centers=5,
    cluster_std=1.2,
    random_state=42
)

# 중심점 초기화 방식 : random
random_model = KMeans(
    n_clusters=5,
    init="random",
    n_init=1,
    random_state=42
)

random_model.fit(X)

print(f"Inertia : {random_model.inertia_}")
print(f"iterations: {random_model.n_iter_}")
print(f"Silhouette Score : {silhouette_score(X, random_model.labels_)}")

# 중심점 초기화 방식 : k-means++
plus_model = KMeans(
    n_clusters=5,
    init="k-means++",
    n_init=1,
    random_state=42
)

plus_model.fit(X)

print(f"Inertia : {plus_model.inertia_}")
print(f"Iterations : {plus_model.n_iter_}")
print(f"Silhouette Score : {silhouette_score(X, plus_model.labels_)}")

# 데이터 프레임으로 비교
comparsion_df = pd.DataFrame({
    "Init" : ["random", "k-means++"],
    "Inertia" : [
        random_model.inertia_,
        plus_model.inertia_
    ]
})
````
