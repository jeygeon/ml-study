### 💡 K-Means
`K-Means`는 데이터를 `K`개의 군집으로 나누고, 각 **군집의 중심점을 반복적으로 갱신하면서 비슷한 데이터끼리 묶는 비지도학습** 알고리즘이다.

이름을 나누어서 해석해보자면
- `K` : 사용자가 설정한 군집의 개수
- `Means` : 군집에 속한 데이터들의 평균
- `Centroid` : 군집에 속한 데이터들의 평균으로 계산한 중심점

> 즉, K개의 중심점을 기준으로 데이터를 묶고 각 그룹의 평균을 이용해 중심점을 계속 갱신한다.

#### 1️⃣ K-Means의 동작원리
1. **K 설정**
    
    생성할 군집의 개수를 결정한다.
2. **Centroid 초기화**
    
    K개의 초기 중심점을 설정한다.
3. **Assignment (데이터 할당)**

    각 데이터를 가장 가까운 중심점의 군집에 할당한다.
4. **Update (중심점 갱신)**

    각 군집에 속한 데이터들의 평균으로 새로운 중심점을 계산한다.
5. **Convergence (수렴)**

    중심점이 더 이상 의미 있게 변하지 않을 때까지 3번과 4번을 반복한다.

K-Means는 한번 데이터를 묶고 끝나는 것이 이나리 마지막으로 설정한 중심점으로부터 더이상 데이터의 군집의 변화가 없을떄까지 학습을 반복한다.

scikit-learn에서는 다음과 같은 파라미터로 반복 학습을 제어한다.
- `max_iter` (기본값: 300) : 한 번의 학습에서 최대 반복 횟수
- `tol` (기본값: 0.00001) : 중심점 변화에 대한 수렴 허용 오차

```python
# scikit-learn을 통한 K-Means 학습
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, y = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=0.7,
    random_state=42
)

# n_clusters : 군집 개수
# init : 초기 중심점 설정 방법
# n_init : 서로 다른 초기 중심점으로 전체 학습을 수행하는 횟수
# max_iter : 한번 학습에서 최대 반복 횟수
# tol : 수렴 판단에 사용하는 허용 오차
model = KMeans(
    n_clusters=3,
    init="k-means++",
    n_init=10,
    max_iter=300,
    tol=0.0001,
    random_state=42
)

model.fit(X)

# labels_ : 각각의 학습 데이터가 어떤 군집으로 편성되었는지 알려준다.
print(model.labels_[:10])

# cluster_centers_ : 학습이 완료된 최종 군집 중심점의 좌표를 반환한다.
print(model.cluster_centers_)

# n_iter_ : 최종 선택된 학습 결과에 실제로 몇번 반복되었는지 확인가능하다.
print(model.n_iter_)

# inertia_ : 각 데이터와 자신이 속한 군집의 중심점 사이의 거리 제곱합이다.
print(model.inertia_)
```

#### ✅ Pandas를 이용한 군집별 데이터 분석
군집화 결과를 DataFrame에 추가하면 각 군집의 특성을 분석하기 용이하다.

```python
# 데이터 프레임 생성
df = pd.DataFrame(X, columns=["feature_1", "feature_2"])

# 군집화 결과값 컬럼 추가
df["cluster"] = model.labels_

print(df.head())

# 군집별 데이터 개수 확인
print(df.groupby("cluster").size())

# 군집별 특성의 평균
print(df.groupby("cluster").mean())
```

#### 실습 문제

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, y = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=0.8,
    random_state=42
)

# 6. 데이터의 shape 출력
print(X.shape)

# 8. KMeans 모델 생성
# n_clusters=4
# init="k-means++"
# n_init=10
# max_iter=300
# random_state=42
model = KMeans(
    n_clusters=4,
    init="k-means++",
    n_init=10,
    max_iter=300,
    random_state=42
)

# 9. 모델 학습
model.fit(X)

# 10. 학습 데이터의 군집화 결과 출력
print(model.labels_[:5])

# 11. 군집 중심점 전체 출력
print(model.cluster_centers_)

# 12. Cluster 0의 중심점 출력
print(model.cluster_centers_[:1])

# 13. 중심점 배열의 shape 출력
print(model.cluster_centers_.shape)

# 14. 실제 반복 학습 횟수 출력
print(model.n_iter_)

# 15. Inertia 출력
print(model.inertia_)

# 16. 군집별 데이터 개수 출력
df = pd.DataFrame(X, columns=["Feature1", "Feature2"])
df["cluster"] = model.labels_
df.groupby("cluster").size()
```

```python
new_data = np.array([
    [2, 3],
    [8, 7],
    [-5, -5],
    [0, 0]
])

# 18. 새로운 데이터의 군집 예측
print(model.predict(new_data))

# 19. 첫 번째 새로운 데이터가 속한 군집 ID 출력
print(model.predict(new_data)[0])

# 20. 첫 번째 새로운 데이터가 속한 군집의 중심점 출력
centroid = model.predict(new_data)[0]
print(model.cluster_centers_[centroid])

# 21. 동일한 하이퍼파라미터로 새로운 KMeans 모델 생성
new_model = KMeans(
    n_clusters=4,
    init="k-means++",
    n_init=10,
    max_iter=300,
    random_state=42
)

# 22. fit_predict()를 이용해 학습 및 군집 결과 반환
print(new_model.fit_predict(X)[:5])

# 23. 기존 model.labels_와 fit_predict() 결과가
# 동일한지 확인
# np.array_equal() 활용
np.array_equal(model.labels_, new_model.fit_predict(X))
```
