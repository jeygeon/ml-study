### 1️⃣ Inertia
`Inertia`란 군집 내 데이터와 중심점 간 거리의 제곱합이다. 이 값이 작을수록 데이터들이 중심점에 가깝게 모여있다는 의미이다.

쉽게 말하면, **Inertia란 각 군집에 속한 데이터들이 중심점을 기준으로 기준으로 얼마나 밀집되어있는지를 나타내는 지표**이다.

`Inertia`의 공식은 아래와 같다.

$Inertia=\sum_{i=1}^{n}\|x_i-\mu_{c_i}\|^2$

- $x_i$ : 개별 데이터
- $\mu_{c_i}$ : 해당 데이터가 속한 군집의 중심점
- $\|x_i-\mu_{c_i}\|^2$ : 데이터와 중심점 사이의 거리의 제곱

예를 들어 어떤 군집의 데이터 3개가 중심점으로부터 각각 1, 2, 3만큼 떨어져있다면

$Inertia=1^2+2^2+3^2=14$

`Inertia` 값는 14가 나오게 된다. 이렇게 모든 군집에 대해 계산한 거리 제곱합을 더한 것이 K-Means의 Inertia이다.

**K-Means는 이러한 Inertia값이 최소화하는 방향으로 학습을 진행하는 알고리즘**이다.

```python
# Inertia값 확인
inertia = model.inertia_
```

`Inertia`값이 무조건 작을수록 군집화가 잘 되었다고 보긴 어렵다.

만약 데이터가 100개가 있고 100개의 군집을 생성한다고 가정하고 모델을 생성했을 시, 모든 데이터가 각각의 군집을 형성하게 되면서 interia의 값은 0이 나오게 된다.

이 경우 interia값은 매우 작게 나오긴 했지만, 의미있는 군집을 형성했다고 보기는 어렵다.

따라서 적절한 군집 개수에서 충분히 작은 interia값을 찾는 것이 중요한데, 이 문제를 해결하기 위해 `Elbow Method`를 사용한다.

### 2️⃣ Elbow Method

elbow method란 군집 개수에 따른 interia값의 변화를 그래프로 나타내고, 팔꿈치처럼 꺾이는 부분을 적절한 K의 후보로 선택하는 방법이다.

````python
# elbow method 확인
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, y = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=0.8,
    random_state=42
)

# inertia값을 넣을 배열 생성
inertias = []

# 군집의 개수 : 1부터 5까지 
k_values = range(1, 6)

for k in k_values:
    model = KMeans(
        n_clusters=k,
        n_init=10,
        random_state=42
    )

    model.fit(X)

    inertias.append(model.inertia_)

# DataFrame으로 결과 확인하기
df = pd.DataFrame({
    "K" : k_values,
    "Inertia" : inertias
})

df
````

위 코드의 결과로 적절한 K값은 4라고 판단할 수 있다.

하지만 모든 데이터셋에서 inertia지점이 명확하게 꺾이는 지점이 보이지 않을수 있다.

그래프의 모든 구간에서 값이 완만하게 감소해서 적절한 K를 판단하기 어려울 수 있다.

이를 보안하기 위해 `Silhouette Score`라는 개념이 나온다.

#### 실습 문제


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, y = make_blobs(
    n_samples=600,
    centers=4,
    cluster_std=0.8,
    random_state=42
)

# 1. n_clusters=2인 KMeans 모델 생성 및 학습
k2_model = KMeans(
    n_clusters=2,
    random_state=42
)
k2_model.fit(X)

# 2. n_clusters=4인 KMeans 모델 생성 및 학습
k4_model = KMeans(
    n_clusters=4,
    random_state=42
)
k4_model.fit(X)

# 3. n_clusters=6인 KMeans 모델 생성 및 학습
k6_model = KMeans(
    n_clusters=6,
    random_state=42
)
k6_model.fit(X)

# 4. 세 모델의 Inertia 출력
print(f"k2 model inertia : {k2_model.inertia_}")
print(f"k4 model inertia : {k4_model.inertia_}")
print(f"k6 model inertia : {k6_model.inertia_}")

# 5. Inertia가 가장 작은 모델 확인
# -> k6 model

# 6. Inertia를 저장할 빈 리스트 생성
inertias = []

# 7. K=1부터 K=10까지 반복하면서 KMeans 모델 학습
# 8. 각 모델의 Inertia를 리스트에 저장
k_range = range(1, 11)
for k in k_range:
    model = KMeans(
        n_clusters=k,
        random_state=42
    )

    model.fit(X)

    inertias.append(model.inertia_)

print(inertias)

# 9. K와 Inertia를 포함하는 DataFrame 생성
# 컬럼명: K, Inertia
df = pd.DataFrame({
    "K" : k_range,
    "Inertia" : inertias
})

# 10. DataFrame 출력
print(df)


# 12. 적절한 K 값 판단
# -> 4

# 13. 선택한 K 값으로 최종 모델 학습
model = KMeans(
    n_clusters=4,
    random_state=42
)

model.fit(X)

# 14. 최종 모델의 Inertia 출력
print(model.inertia_)

# 15. 최종 모델의 군집별 데이터 개수 확인
new_df = pd.DataFrame(X, columns=["feature1", "feature2"])
new_df["cluster"] = model.labels_
print(new_df.groupby("cluster").size())
````
