#### 💡 로지스틱 회귀 모델
로지스틱 회귀 모델은 입력 데이터와 정답을 이용해서 가중치(w)와 절편(b)을 학습한다.

$z = wx + b$

로지스틱 회귀는 이 선형 점수를 시그모이드 함수에 넣어 클래스 1에 속할 확률을 예측한다.

```python
# 모델 학습
import numpy as np
from sklearn.linear_model import LogisticRegression

# 공부 시간
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# 합격 여부
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

print(f"가중치 : {model.coef_}")
print(f"절편 : {model.intercept_}")
```

```python
# 여러 개의 변수로 모델 학습
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# 공부 시간, 출석률
X = np.array([
    [1, 50],
    [2, 60],
    [3, 65],
    [3, 85],
    [4, 75],
    [5, 90],
    [6, 95],
    [2, 40]
])

# 합격 여부
y = np.array([0, 0, 0, 1, 1, 1, 1, 0])

# 스케일링 + 로지스틱 회귀
model = make_pipeline(
    StandardScaler(),
    LogisticRegression()
)

model.fit(X, y)

new_student = np.array([[4, 80]])

y_pred = model.predict(new_student)
print(y_pred)

y_prob = model.predict_proba(new_student)
print(y_prob)
```

#### 실습문제

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# 하루 운동 시간
X = np.array([0, 1, 2, 3, 4, 5, 6, 7]).reshape(-1, 1)

# 목표 달성 여부
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

# 1. 학습된 가중치 출력
print(f"학습된 모델 가중치 : {model.coef_}")

# 2. 학습된 절편 출력
print(f"학습된 모델 절편 : {model.intercept_}")

# 3. 가중치가 양수인지 음수인지 확인
# -> 양수

# 4. 가중치의 부호가 의미하는 바를 주석으로 설명
# -> 값이 증가할수록 클래스1일 확률이 증가한다.

# 새로운 운동 시간
X_new = np.array([[2], [4], [6]])

# 1. 각 운동 시간의 예측 클래스 출력
y_pred = model.predict(X_new)
print(y_pred)

# 2. 각 운동 시간의 클래스 1 예측 확률 출력
print(model.predict_proba(X_new)[:, 1])

# 3. 운동 시간이 증가할수록 목표 달성 확률이 어떻게 변하는지 설명
# -> 운동시간이 증가할수록 목표달성 확률이 증가한다. 가중치가 양수이기 때문
```
