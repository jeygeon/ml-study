### 💡 Odds & Log-Odds

#### 1️⃣ Odds(오즈)
Odds란 어떤 사건이 발생할 확률과 발생하지 않을 확률의 비율이다.

예를들어 시험 합격 확률이 80%라고 생각해봤을 때
- 합격할 확률 : 80%
- 불합격할 확률 : 20%

오즈는 다음과 같다.

$\text{Odds}=\frac{p}{1-p}$

여기서 $p$는 합격할 확률이고 오즈를 계산해봤을 때 값은 4가 나온다.

즉, **합격할 가능성이 불합격할 가능성의 4배라는 의미**이다.

#### 2️⃣ Log-Odds(로그 오즈)

로그오즈는 로즈에 자연로그를 적용한 값이다.

$\text{Log-Odds}=\ln\left(\frac{p}{1-p}\right)$

로그를 사용하는 이유는 오즈는 0보다 큰 값이지만, 로그오즈는 음수부터 양수까지 모든 실수값을 표현할 수 있기 때문이다.

예를들어 합격확률이 80%라면,

$\text{Odds}=4$

$\text{Log-Odds}=\ln(4)\approx1.386$

반대로 합격확률이 20%라면,

$\text{Odds}=0.25$

$\text{Log-Odds}=\ln(0.25)\approx-1.386$

라는 결과가 나온다.

로지스틱 회귀의 기본 원리를 정리하자면 이렇게 정리가 가능하다.

1. 입력 데이터를 입력
2. 로그오즈 값을 얻는다.
3. 로그오즈값을 시그모이드 함수로 변환해서 0과 1사이의 확률을 얻는다.
4. 해당 확률로 클래스를 결정한다.

```python
# python에서 로그오즈 확인하기
import numpy as np
from sklearn.linear_model import LogisticRegression

# 공부 시간
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# 합격 여부
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

X_new = np.array([[2], [3], [4], [5]])

# 로그오즈값 확인
# decision_function() : 시그모이드 함수를 적용하기 전의 z값인 로그오즈를 반환한다.
scores = model.decision_function(X_new)
print(scores)

# 클래스별 예측 확률 출력
prob = model.predict_proba(X_new)
print(prob)
```
