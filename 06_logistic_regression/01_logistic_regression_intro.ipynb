### 💡 Logistic Regression
로지스틱 회귀란 **입력 데이터를 이용해 특정 클래스에 속할 확률을 예측**하고, 그 확률을 바탕으로 클래스를 분류하는 머신러닝 알고리즘이다.

예를들어 공부시간에 따른 합격 여부를 예측하고 싶을 때 사용한다.

예측하고자 하는 것은 시험점수가 아니라 합격여부이다.

이렇게 두 개의 클래스로 구분하는 문제를 **이진분류**(Binary Classification)이라고 한다.

로지스틱 회귀에서는 **시그모이드**(Sigmoid)함수를 이용해서 출력을 0과 1 사이로 변환한다.

#### ✅ 로지스틱 회귀의 동작 원리
로지스틱 회귀는 크게 세 단계를 거친다.

1. 선형 점수 계산
    - $z=wx+b$
    - 입력값을 이용해 점수를 계산한다.

2. 시그모이드 함수 적용
    - $p=\frac{1}{1+e^{-z}}$
    - 점수를 0~1 사이의 확률로 변환

3. 클래스 결정
    - $p\geq0.5\Rightarrow1$, $p<0.5\Rightarrow0$
    - 기본 임계값 0.5를 기준으로 분류

예를 들어 합격 확률이 `0.87`이면 **합격**(1), `0.23`이면 **불합격**(0)으로 예측하는 것이다.

```python
# 로지스틱 회귀 모델 만들어보기
import numpy as np

from sklearn.linear_model import LogisticRegression

# 공부 시간
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# 합격 여부
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()

model.fit(X, y)

X_new = np.array([[2.5]])

y_pred = model.predict(X_new)

print(y_pred)
```

```python
# 로지스틱 회귀에서는 predict_proba()를 이용해 확률을 예측할 수 있다.
# [클래스0 확률, 클래스1 확률]
prob = model.predict_proba(X_new)

print(prob)
```

#### 실습문제

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# 하루 운동 시간
X = np.array([0, 1, 2, 3, 4, 5]).reshape(-1, 1)

# 목표 달성 여부
y = np.array([0, 0, 0, 1, 1, 1])

# 1. LogisticRegression 모델 생성
model = LogisticRegression()

# 2. fit()으로 모델 학습
model.fit(X, y)

# 3. 운동 시간이 1.5시간, 3.5시간인 경우 결과 예측
X_new = np.array([[1.5], [3.5]])
print(model.predict(X_new))

# 4. 각 데이터의 클래스별 예측 확률 출력
print(model.predict_proba(X_new))

# 개념 확인
# 1. 로지스틱 회귀는 회귀 문제와 분류 문제 중 주로 어디에 사용될까?
# -> 클래스 예측이라 분류 문제에 사용한다.

# 2. 선형회귀의 출력값을 그대로 확률로 사용하기 어려운 이유는 무엇일까?
# -> 선형회귀는 예측값의 제한이 없어서 음수나 1보다 큰 값이 나와 확률로 해석하기 어렵기 때문이다.

# 3. predict()와 predict_proba()의 차이는 무엇일까?
# -> predict() : 클래스 출력
# -> predict_proba() : 해당 클래스일 확률 출력

# 4. 클래스 1의 예측 확률이 0.8이고 임계값이 0.5라면 어떤 클래스로 분류될까?
# -> 클래스 1일 확률로 분류
```
