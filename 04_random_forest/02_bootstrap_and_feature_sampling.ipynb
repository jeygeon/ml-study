#### 1️⃣ Bootstrap Sampling

Bootstrap Sampling은 원본 데이터에서 복원 추출하여 새로운 학습 데이터셋을 만드는 과정이다.

복원 추출이기 때문에 같은 데이터가 여러번 선택 될 수 있다.

```python
# np.random.choice() : 복원 추출을 구현할 수 있다.
import numpy as np

data = np.array(["A", "B", "C", "D", "E"])

np.random.seed(42)

# replace=True : 복원 추출, False : 비복원추출
sample = np.random.choice(
    data,
    size=5,
    replace=True
)

print(sample)
```

```python
# 여러개의 Bootstrap 데이터셋 만들기
np.random.seed(42)

for i in range(3):
    sample = np.random.choice(
        data,
        size=len(data),
        replace=True
    )

    print(f"Tree {i} : {sample}")
```

#### 2️⃣ Random Feature Sampling

Random Feature Selection은 각 노드에서 분할할 때 전체 Feature 중 일부를 무작위로 선택하는 방법이다.

실제로 scikit-learn에서는 직접 Sampling을 구현할 필요는 없다.

```python
from sklearn.ensemble import RandomForestClassifier

# n_estimators=100 : 트리 100개 생성
# bootstrap=True : 각 트리의 학습 데이터 복원 추출
# max_features="sqrt" : 각 분할에서 전체 Feature 개수의 제곱근에 해당하는 개수를 후보로 선택
model = RandomForestClassifier(
    n_estimators=100,
    bootstrap=True,
    max_features="sqrt",
    random_state=42
)
```
