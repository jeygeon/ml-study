### 💡 Gini & Entropy

`Decision Tree`는 **데이터의 불순도를 감소시키는 방향으로 노드를 분할**한다.

#### 1️⃣ 불순도 (Impurity)

불순도란 하나의 노드에 서로 다른 클래스가 얼마나 섞여있는지를 나타내는 개념이다.

예를들어 초보자, 숙련자 두 가지 클래스를 가진 데이터가 있고 노드가 3개 존재한다고 가정해보자.

각각의 노드엔 **초보자**와 **숙련자** 비율이 `10명/0명`, `8명/2명`, `5명/5명`이 있다고 한다면, 불순도는 순서대로

`불순도 없음`, `불순도 낮음`, `불순도 높음`이라고 판단할 수 있다.

| 노드 | 초보자:숙련자 비율 | 불순도 |
| --- | --- | --- |
| A | 10:0 | 없음 |
| B | 8:2 | 낮음 |
| C | 5:5 | 높음 |

불순도의 특징을 정리해보자면 아래와 같다.
- 하나의 클래스만 있다면 불순도는 0이다.
- 데이터가 전체 갯수보다는 하나의 노드에 존재하는 클래스의 비율이 중요하다.
- 여러 클래스가 균등하게 있을수록 불순도는 높다.
- 최상위 노드인 Root node라고 불순도가 가장 높은건 아니다.

> 중간 한줄 정리 : Decision Tree는 불순도를 감소시키는 방향으로 노드를 분할하는데, 불순도를 측정하는 방법에는 `Gini`와 `Entropy`가 있다.

#### 2️⃣ Gini (지니)

`Gini`는 하나의 노드에서 데이터를 무작위로 두 번 추출했을 때 서로 다른 클래스가 나올 확률을 의미한다.

복원 추출이란 데이터를 하나 뽑은 후 다시 넣고 다음 데이터를 뽑는 방식이다.

$Gini = 1 - \sum_{i=1}^{C}p_i^2$

- $C$ : 전체 클래스의 개수
- $p_i$ : 특정 클래스가 나올 확률
- $p_i^2$ : 특정 클래스가 연속으로 두 번 나올 확률
- $\sum_{i=1}^{C}p_i^2$ : 같은 클래스가 나올 전체 확률
- $1 - \sum_{i=1}^{C}p_i^2$ : 서로 다른 클래스가 나올 확률

같은 클래스가 연속으로 두 번 나올 확률을 구하기 위해 제곱을 하는 것이고, 전체확률에서 같은 클래스가 나올 확률을 빼면 서로 다른 클래스가 나올 확률을 구할 수 있기 때문이다.


```python
# 파이썬 코드로 확인해보기
beginner = 8
experienced = 2

total = beginner + experienced

# p0 : 초보자가 나올 확률
# p1 : 숙련자가 나올 확률
p0 = beginner / total
p1 = experienced / total

# 전체에서 두 번 뽑을때 동일한 클래스가 낭로 확률을 빼준다.
gini = 1 - (p0 ** 2 + p1 ** 2)
# round(gini, 4) : gini에 저장된 값을 소수점 넷째 자리까지 반올림하여 출력한다.
print(round(gini, 4))
```

```python
# scikit-learn에서 Gini 사용해보기
from sklearn.tree import DecisionTreeClassifier

# Gini 사용 (디폴트)
gini_model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)
```

#### 3️⃣ Entropy (엔트로피)

`Gini`와 마찬가지로 `Entropy`도 데이터의 불확실성(불순도)을 측정하는 지표이다.

`Gini`가 **서로 다른 클래스가 나올 확률을 계산**했다면, `Entropy`는 **어떤 클래스가 나올지 얼마나 예측하기 어려운지를 계산**한다.

`Entropy`를 알기전에 정보량이라는 개념을 알아야 한다.

**정보량**(Information)이란 특정 사건이 발생했을 때 얻는 정보의 크기를 말한다.

예를들어 초보자가 90%, 숙련자가 10%인 노드에서 데이터를 하나 뽑는다고 했을 때

초보자가 나올 확률은 90% 이며 얻을 수 있는 정보량은 낮다.

숙련자가 나올 확률은 10% 이며 얻을 수 있는 정보량은 높다.

**발생 확률이 낮은 사건일수록 예측하기 어려운 결과이기 때문에 얻는 정보량이 커진다.**

정보량 계산 공식은

$Infromation = -\log_2 (p)$

- $p$ : 사건의 발생 확률
- $log_2$ : 밑이 2인 로그
- 마이너스(-) : 확률을 로그값에 넣으면 음수가 나오기 때문에 양수로 변환을 위해 사용

`Entropy`의 계산 공식이다.

$Entropy = - \sum_{i=1}^{C}p_i\log_2 (p_i)$

`Entropy`는 위에서 계산한 **정보량**에 **사건이 발생할 확률**만 곱해주면 된다.
> 정보량 * 사건이 발생할 확률

```python
# 파이썬 코드로 확인해보기
import numpy as np

beginner = 8
experienced = 2

total = beginner + experienced

# 클래스별 확률
p0 = beginner / total
p1 = experienced / total

# 클래스별 정보량
info0 = -np.log2(p0)
info1 = -np.log2(p1)

# Entropy 게산
entropy = (p0 * info0) + (p1 * info1)

print(round(entropy, 4))
```

```python
# scikit-learn에서 Entropy 사용해보기
from sklearn.tree import DecisionTreeClassifier

# 디폴트 값은 gini이기 때문에 꼭 하이퍼파라미터를 선언해주어야 한다.
entropy_model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)
```
