### 1️⃣ 머신러닝이란?

머신러닝은 다음과 같이 설명할 수 있다.

> 컴퓨터가 데이터를 통해 패턴을 발견하고, 그 패턴을 이용해 새로운 데이터에 대한 예측이나 결정을 수행하는 과정.

예를들어 중고차 가격을 예측하고 싶을 때 가격예측을 개발자가 직접 규칙을 만들어서 할 수 있다. **ex) 사용기간이 1년 증가할 때마다 가격이 200만원씩 감소**

하지만 현실의 중고차 가격은 사용 기간 뿐 아니라 주행거리, 사고이력, 차종 등 다양한 요인의 영향을 받는다.

중요한건 머신러닝이 모든 것을 스스로 이해하는게 아니라, **데이터에 잘 맞는 수학적 규칙을 찾아가는 과정**이라는 점이다.

### 2️⃣ Feature(X), Target(y), Model 이해하기

머신러닝의 핵심은 다음과 같다.

> 입력 X를 모델 f에 넣으면 예측값 ŷ이 나온다.

여기서 용어는 다음과 같이 구분한다.

| 용어 | 의미 | 중고차 가격 예측 예시 |
| --- | --- | --- |
| Feature (특성) | 예측에 사용하는 입력 변수 | 사용거리, 주행시간 |
| Target (목표변수) | 예측하려는 실제 정답 | 실제 판매 가격 |
| Model (모델) | 입력으로부터 예측값을 만드는 규칙 | 가격 예측 함수 |
| Prediction (예측값) | 모델이 계산한 결과 | 2,400만원 |
| Parameter (매개변수) | 모델이 학습한 내부 값 | 감가 계수 등 |

주의할 점은 `y`는 실제 정답을, `ŷ`는 모델의 예측값을 나타내는 경우가 많다. 항상 두 값이 같지는 않다.

### 3️⃣ 모델이 학습한다는 것은 무슨 의미일까

모델 학습이란 아래 예시로 설명할 수 있다.

예를들어 공부시간이 각각 1, 2, 3시간인 학생의 시험점수가 각각 50, 60, 70점이라고 했을 때

이 관계를 수식으로 나타내면 $\hat{y} = 10X + 40$ 이렇게 나타낼 수 있다.

- $X$ : 공부시간
- 10 : 공부시간이 1시간 늘어날 때 예상 점수가 증가하는 양
- 40 : 직선의 절편
- $\hat{y}$ : 예상시험 점수

우리는 10과 40이라는 숫자가 직접 계산을 하면서 알아냈지만, 머신러닝에서는 알고리즘이 데이터에 잘 맞는 값을 찾아낼 수 있다.

이것이 **모델 학습(Training)** 이다.

```python
# python으로 첫 번째 머신러닝 모델 만들기
# 10X + 40이라는 위의 공식을 직접 계산하지않고,
# 머신러닝 라이브러리가 직접 학습해보도록 테스트

# 공부시간이 증가할수록 시험점수가 증가하는 직선형태를 보이기때문에
# LinearRegression 모델 사용
import numpy as np
from sklearn.linear_model import LinearRegression

# 공부 시간 (Feature)
# scikit-learn에서는 대부분의 머신러닝 모델은 입력 데이터 X를 2차원형태로 받는다.
# 공부시간뿐 아니라 고려하는 변수가 많아질수록 값 추가하기 용이하다.
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

# 시험 점수 (Target)
# scikit-learn에서는 일반적인 단일 출력 회귀분류에서는 정답을 1차원 배열로 전달한다.
y = np.array([50, 60, 70, 80, 90])

# 모델학습
# LinearRegression()의 fit()은 데이터에 맞는 직선의 개수와 절편을 학습한다.
# fit(X, y) : 데이터와 정답을 사용해 규칙 학습
model = LinearRegression()
model.fit(X, y)

# 새로운 데이터
new_X = np.array([[6]])

# 새로운 데이터로 결과 예측
# predict(X) : 학습된 규칙을 사용해 새로운 입력의 결과를 예측
# 반환값은 numpy 배열
prediction = model.predict(new_X)

print("예상 점수:", prediction[0])
```

### 4️⃣ 모델은 어떻게 좋은 규칙을 판단할까

만약 실제 시험점수가 80점인데 모델은 75점이라고 예측했다면? 또는 모델이 85점이라고 예측했다면?

두 예측 모두 실제 점수와 5점 차이가 나지만 오차의 부호는 다르다.

때문에 모든 오차를 제곱해서 합산하는 방법을 사용할 수 있다.

`SSE(Sum of Squared Errors)`, 즉 오차의 제곱합이다.

```python
# SSE 예제
actual = np.array([50, 60, 70])
predicted = np.array([55, 58, 72])

errors = actual - predicted

sse = np.sum(errors ** 2)

# 오차의 제곱합
print(sse)
```

이렇게 예측값과 정답의 차이를 수치화하면 모델의 예측이 얼마나 틀렸는지 판단할 수 있다.

**손실함수**(Loss Function)는 모델의 예측이 얼마나 잘못되었는지 측정하는 기준이다.

모델학습에서는 이러한 손실을 줄이는 방향으로 규칙을 찾는 경우가 많다.
