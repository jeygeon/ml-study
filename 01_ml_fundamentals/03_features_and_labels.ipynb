### 💡 Features and Labels

머신러닝은 다음과 같은 관계를 설명한다.

$f : X \rightarrow Y$

입력 X를 모델에 전달하면 모델이 출력 Y를 예측하는 구조이다.

예를들어 회사에서 직원의 연봉을 예측하는 모델을 만든다고 생각했을 때

다음과 같이 용어를 나눌 수 있다.

- Feature(X) : 경력, 나이, 자격증 수 - 예측에 사용되는 정보
- Target(y) : 연봉 - 예측하려는 실제 정답
- Model(f) : 입력 정보로 연봉을 예측하려는 규칙

Feature은 여러개를 함께 사용하는 경우가 많다.

예를들어 연봉을 예측하려고 하면 경력뿐만 아니라 위에서 처럼 자격증 갯수, 나이 등을 함께 고려할 때 더 많은 정보를 제공할 수 있을 것이다.

하지만 특성이 많다고 무조건 예측 성능이 좋아지는 것은 아니며, 관련 없는 특성이 많아지면 오히려 모델 성능이 떨어질 수 있다.

Feature의 자료형 또한 중요하다.

예를들어 아래와 같은 데이터를 볼 때

```python
# resigned 0 : 퇴사하지않음. 1 : 퇴사함
data = {
    "age": [25, 30, 35, 40],
    "salary": [3000, 4000, 5000, 6000],
    "department": ["IT", "HR", "IT", "Sales"],
    "resigned": [0, 0, 1, 1]
}
```

여기서 resigned은 숫자로 저장되어있긴 하지만 분류 문제의 정답이다.

데이터 타입이 숫자라고 해서 반드시 회귀 문제인 것은 아니다.

그리고 department처럼 문자열로 이루어진 범주형 Feature은 대부분의 scikit-learn 모델에 그대로 적용할 수 없어서, 숫자로 변환하는 인코딩 과정이 필요하다.

````python
# 실제 모델에 다중 Feature 전달해보기
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "experience": [1, 2, 3, 4, 5, 6],
    "age": [24, 26, 28, 30, 32, 34],
    "certificates": [1, 1, 2, 2, 3, 3],
    "salary": [3000, 3300, 3900, 4200, 4800, 5100]
}

df = pd.DataFrame(data)

X = df[["experience", "age", "certificates"]]
y = df["salary"]

# 모델 객체 생성
model = LinearRegression()

# 모델 학습
model.fit(X, y)

# 모델 테스트
# 경력 10년, 나이 50, 자격증 수 5개의 사람의 연봉 예측
y_pred = model.predict([[10, 50, 5]])
print(y_pred[0])
````

### 실습

````python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression

data = {
    "customer_id": [101, 102, 103, 104, 105, 106],
    "age": [25, 32, 28, 45, 38, 29],
    "monthly_spending": [50000, 120000, 80000, 200000, 150000, 70000],
    "visits": [3, 8, 5, 12, 9, 4],
    "membership_years": [1, 3, 2, 5, 4, 1],
    "next_month_spending": [60000, 130000, 90000, 220000, 160000, 75000]
}

df = pd.DataFrame(data)

# 목표: 고객의 다음 달 소비 금액을 예측하는 모델 만들기
# 1. 전체 데이터의 행과 열 개수를 확인해 봐.
print(df.shape)

# 2. next_month_spending을 Target(y)으로 분리해 봐.
y = df["next_month_spending"]

# 3. age, monthly_spending, visits, membership_years를 Feature(X)로 분리해 봐.
X = df[["age", "monthly_spending", "visits", "membership_years"]]

# 4. X와 y의 shape을 출력해 봐.
print(X.shape)
print(y.shape)

# 5. customer_id를 일반적으로 Feature에서 제외하는 이유를 설명해 봐.
# 아이디는 인덱스값인데 인덱스값으로는 일반적으로 우리가 알아내고자하는 데이터에 관여하지 않는다.

# 6. LinearRegression 모델을 생성하고 전체 데이터로 학습하기
model = LinearRegression()
model.fit(X, y)

# 7. 새로운 고객 데이터 생성하기
new_customer = pd.DataFrame({
    "age" : [50],
    "monthly_spending" : [200000],
    "visits" : [10],
    "membership_years" : [1]
})

# 8. 새로운 고객의 다음 달 소비 금액 예측하기
y_pred = model.predict(new_customer)

# 9. prediction.shape을 출력하고 의미 설명하기
print(y_pred.shape)
# 고객 한명의 예측 값 1개 이기때문이다

# 10. 고객을 2명으로 늘리면 입력 데이터의 shape이 어떻게 달라지는지 설명하기
# 고객이 한명일 때는 입력 데이터가 (1, 4)이였다면 두명일 때는 (2, 4)으로 바뀐다.
````
