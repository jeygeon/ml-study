### 1️⃣ 지도학습(Supervised Learning)이란

`지도학습`이란 **문제와 정답(Label)이 함께 있는 데이터를 제공하여 학습시키는 방식**이다.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5],
    "score": [50, 60, 70, 80, 90]
})

X = df[["study_hours"]]
y = df["score"]

model = LinearRegression()
model.fit(X, y)

new_student = pd.DataFrame({
    "study_hours": [6]
})

prediction = model.predict(new_student)

print("예상 시험 점수:", prediction[0])
```

지금까지 학습 했던 모든 것이 다 지도학습이다.

`LinearRegression` 모델은 지도학습 모델로써 공부 시간과 실제 점수의 관계를 학습하기 위해 정답 y를 사용한다.

### 2️⃣ 비지도학습(Unsupervised Learning)이란

`비지도학습`은 정답없이 **데이터 내부의 숨겨진 패턴이나 구조를 찾아내는 학습 방식**이다.

예를들어 아래와 같은 온라인 쇼핑몰 고객 데이터가 있다고 해보자.

| 고객 | 월 소비 금액 | 월 방문 횟수 |
| --- | --- | --- |
| A | 20,000원 | 2회 |
| B | 25,000원 | 3회 |
| C | 200,000원 | 20회 |
| D | 220,000원 | 22회 |
| E | 70,000원 | 7회 |
| F | 75,000원 | 8회 |

이 데이터에는 고객이 어떤 그룹에 속하는지 정답은 없다.

하지만 데이터를 살펴보면 소비 금액과 방문 횟수가 비슷한 고객들이 분명 존재한다.

비지도학습은 이런 데이터의 구조를 발견하는데 활용할 수 있다.

비지도학습의 대표적인 예로는 **군집화(Clustering)** 가 있다.

````python
# 비지도학습 구현
# KMeans : 비지도학습의 대표적인 알고리즘
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

df = pd.DataFrame({
    "monthly_spending": [20000, 25000, 70000, 75000, 200000, 220000],
    "visits": [2, 3, 7, 8, 20, 22]
})

# KMeans는 데이터 사이의 거리를 이용하는 알고리즘이다.
# 때문에 스케일링이 필요하다.

# 스케일러 생성
scaler = StandardScaler()

# 스케일링
X = df[["monthly_spending", "visits"]]
X_scaled = scaler.fit_transform(X)

# 비지도학습 진행
# KMeans() : 군집화 모델 생성
# n_cluster=3 : 군집을 3개로 나누도록 지정
# n_init=10 : 서로 다른 초기 중심점으로 10번 실행해 가장 좋은 결과 선택
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# 모델 학습
model.fit(X_scaled)

# labels_ : 학습 데이터의 군집 번호
print(model.labels_)

# 군집화 데이터를 데이터프레임 컬럼 추가
df["cluster"] = model.labels_

print(df)
````

### 실습

````python
import pandas as pd

from sklearn.linear_model import LinearRegression

data = {
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "score": [45, 50, 58, 63, 70, 76, 84, 91]
}

df = pd.DataFrame(data)

# A. 지도학습
# study_hours, attendance를 Feature(X)로 분리해 봐.
X = df[["study_hours", "attendance"]]

# score를 Target(y)으로 분리해 봐.
y = df["score"]

# LinearRegression 모델을 생성하고 학습시켜 봐.
model = LinearRegression()
model.fit(X, y)

# 공부 시간 9시간, 출석률 98인 학생의 점수를 예측해 봐.
y_pred = model.predict(pd.DataFrame({
    "study_hours" : [9],
    "attendance" : [98]
}))
print(y_pred)

# 이 모델이 지도학습인 이유를 설명해 봐.
# 학습데이터와 정답이 같이 제공되기 때문이다.
````

````python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

data = {
    "monthly_spending": [
        20000, 25000, 30000,
        80000, 90000, 100000,
        200000, 220000, 240000
    ],
    "visits": [
        2, 3, 3,
        7, 8, 9,
        18, 20, 22
    ]
}

df = pd.DataFrame(data)

# B. 비지도학습
# 1. 두 컬럼을 Feature(X)로 분리하고 X.shape 출력하기
X = df[["monthly_spending", "visits"]]
print(X.shape)

# 2. StandardScaler를 사용해 X 표준화하기
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. KMeans로 군집 3개 생성하기 (random_state=42, n_init=10)
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# 4. 각 고객의 군집 번호를 cluster 컬럼에 추가하기
cluster = model.fit(X_scaled)
df["cluster"] = cluster.labels_

# 5. KMeans가 비지도학습인 이유 설명하기
# 정답이 없고 주어진 데이터를 사용해서 그룹을 나누기 때문
````
